# FMCG stockout EDA
Run from repository root after generating data. Synthetic educational data only.

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv('model_features.csv', parse_dates=['date'])
print(df.shape)
print(df.isna().sum())
print(df.duplicated(['date','sku','warehouse_id']).sum())
df.describe(include='all')
```

```python
df.groupby('date')['stockout_within_7d'].mean().plot(title='Future stock-depletion prevalence over time')
plt.ylabel('Positive fraction')
plt.show()
```

```python
df.groupby('category')['stockout_within_7d'].agg(['count','mean'])
```

```python
df[['on_hand_qty','days_of_supply','avg_sales_30d']].hist(figsize=(10,4))
plt.tight_layout()
plt.show()
```

## Interpretation
Compare class prevalence across time, warehouses and categories. Fulfilled sales are censored by availability. Do not select features using final-test outcomes. Exploratory analysis does not demonstrate causal business impact.
